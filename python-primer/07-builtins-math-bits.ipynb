## Built-ins that replace loops

````python
nums, words = [4, -2, 9], ["kiwi", "fig", "banana"]
print(sum(nums), min(nums), max(nums))   # O(n); min/max of empty raise ValueError
print(max([], default=0))                # safe on empty
print(min(words, key=len))
print(any(x < 0 for x in nums))          # short-circuits
print(all(c.isdigit() for c in "123"))
print(sum(1 for x in nums if x > 0))     # count matching
print(divmod(17, 5))                     # (3, 2)
print(abs(-3))
````

`map(int, parts)` and generator expressions (`x * x for x in nums`) are lazy; wrap in `list()` if you need to index.

## itertools

````python
from itertools import accumulate, combinations, groupby, pairwise, permutations, product

nums = [3, 1, 4, 1, 5]
print(list(accumulate([1, 2, 3])))              # running sums
print(list(accumulate([1, 2, 3], initial=0)))   # prefix-sum array
print(list(accumulate(nums, max)))              # running max
print(list(combinations([1, 2, 3], 2)))
print(list(permutations([1, 2, 3])))            # all 6 orderings
print(list(product("ab", repeat=2)))
print([(k, len(list(g))) for k, g in groupby("aaabcc")])
print(list(pairwise([1, 2, 4])))                # adjacent pairs
````

In an interview, using `permutations` for a Permutations problem misses the point; the interviewer wants the backtracking. Use these for the parts of a solution that aren't the point.

## Bits

| Expression | Meaning |
|---|---|
| `x & 1` | 1 if odd |
| `x >> 1`, `x << 1` | halve (floor), double |
| `x & (x - 1)` | x with its lowest set bit cleared |
| `x & -x` | only the lowest set bit |
| `x > 0 and x & (x - 1) == 0` | power of two |
| `a ^ a == 0`, `a ^ 0 == a` | XOR cancels pairs (Single Number) |
| `bin(x).count("1")` or `x.bit_count()` | number of 1 bits |
| `(x >> i) & 1` | bit i |
| `x | (1 << i)` | set bit i |

**The Python-specific trap:** integers have unlimited precision, so negative numbers behave as if they have infinitely many leading 1s. `~5 == -6`. `-1 >> 1 == -1` forever. Problems that assume 32-bit integers (Reverse Bits, Sum of Two Integers, Number of 1 Bits on a negative) need a mask:

````python
MASK = 0xFFFFFFFF
x = -5                        # pretend this came from a 32-bit problem
x &= MASK                     # keep only the low 32 bits
print(x, hex(x))
if x > 0x7FFFFFFF:            # convert back to a signed 32-bit value
    x = ~(x ^ MASK)
print(x)
````

````python
x = 12                            # 0b1100
print(bin(x), x & 1, x >> 1, x << 1)
print(bin(x & (x - 1)))           # lowest set bit cleared
print(bin(x & -x))                # only the lowest set bit
print(16 > 0 and 16 & 15 == 0)    # power of two
print(5 ^ 3 ^ 5)                  # XOR cancels pairs
print(bin(13).count("1"), (13).bit_count())
print((13 >> 2) & 1, bin(13 | (1 << 1)))   # read bit 2, set bit 1
print(~5, -1 >> 1)                # negatives have infinite leading 1s
````
